## Task 1: Create a database using SQLite

```python
import sqlite3

```

```python
#connecting to db
conn =sqlite3.connect('INSTRUCTOR.db')
```

Cursor class is an instance using which you can invoke methods that execute SQLite statements, fetch data from the results sets of the queries.

```python
#CURSOR OBJECT
cursor_obj=conn.cursor()
```

```python
#creating a table in the database
#check if the atbel exists and drop it
cursor_obj.execute("DROP TABLE IF EXISTS instructor")
```

## Task 2: Create a table

```python
#creating the table
table="""CREATE TABLE IF NOT EXISTS INSTRUCTOR(ID INTEGER PRIMARY KEY NOT NULL, FNAME VARCHAR(20), LNAME VARCHAR(20), CITY VARCHAR(20), CCODE CHAR(2));"""

cursor_obj.execute(table)

print("table is ready")
```

## Task 3: Insert Data into the table

```python
#inserting data into the db
cursor_obj.execute('''insert into INSTRUCTOR values(1, 'Edwards','George','NAIROBI','KE'),(2,'Barack','Obama','WASHINGTON DC','US'),(3,'Vladamir','Putin','MOSCOW','RU') ''')
```

## Task 4: Query data in the table

```python
#Fetch few rows from the table
statement='''SELECT * FROM INSTRUCTOR '''
cursor_obj.execute(statement)
print("All the data")

output_all=cursor_obj.fetchall()
for row_all in output_all:
    print(row_all)
```

```python
#fetch fname only
statement='''SELECT FNAME FROM INSTRUCTOR '''
cursor_obj.execute(statement)
print("First names only")

fname_all=cursor_obj.fetchall()
for fname_rows in fname_all:
    print(fname_rows)
```

```python
#update Edwards city to New York and countrycode to US
statement='''update INSTRUCTOR set CITY="NEW YORK", CCODE="US" where FNAME="Edwards" '''
cursor_obj.execute(statement)
print("Updated Table")
query=''' SELECT * FROM INSTRUCTOR '''
cursor_obj.execute(query)
new_table=cursor_obj.fetchmany(1)
for new_rows in new_table:
    print(new_rows)
```
