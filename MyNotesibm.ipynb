### Lists and tuples


```python
#Tuples
a=(1,2,3,"dim","pass",(1,2,3,4))
print(f"This is a tuple {a}")
#Tuples are immutable
try:
    a[0]=4
    print(a)
except:
    print("It is Unmuttable")
    print(a[0])
c=len(a)-1
print(a[c][3])
#accessing items in a list using negative indexing. the last item is always -1 counting backwards towards the left
#code below accesses the second item in the tuple inside the tuple
print(a[-1][-3])
#Lists
b=[1,2,3,"dim","pass",[1,2,3,4]]
print(f"This is a list {b}")
#lists are mutable
b.extend([5,6])
print(b)
c=b
#since b=c when we change an element in c b should change too
c[0]="hi"
print(b)
print(c)
#cloning a list
d=b[:]
print (d)
# Now when d is changed it does not affect b or c
d[-len(d)]="hello"
print (d)
```

```python
class Circle(object):

    def __init__(self, radius=0, color='blue'):
        self.radius=radius
        self.color=color
    def add_radius(self, r):
        
        self.radius=self.radius+r
        return(self.radius)
    def drawCircle(self):
        
        plt.gca().add_patch(plt.Circle((0,0), radius=self.radius, fc=self.color))
        plt.axis('scaled')
        plt.show
```

### TextAnalysis

```python
givenString="But I must explain to you how all this mistaken idea of denouncing pleasure and praising pain was born and I will give you a complete account of the system, and expound the actual teachings of the great explorer of the truth, the master-builder of human happiness. No one rejects, dislikes, or avoids pleasure itself, because it is pleasure, but because those who do not know how to pursue pleasure rationally encounter consequences that are extremely painful. Nor again is there anyone who loves or pursues or desires to obtain pain of itself, because it is pain, but because occasionally circumstances occur in which toil and pain can procure him some great pleasure. To take a trivial example, which of us ever undertakes laborious physical exercise, except to obtain some advantage from it? But who has any right to find fault with a man who chooses to enjoy a pleasure that has no annoying consequences, or one who avoids a pain that produces no resultant pleasure?"
```

```python
class textAnalyzer(object):
    def __init__(self,text):
        lCaseText=self.text.lower()#change all to lowercase
        #remove all punctuation
        NoPuncText=lCaseText.replace(".","").replace("?","").replace(",","").replace("!","")
        self.fmttext=NoPuncText

    def freqAll(self):
        #split to form a list of all words
        listText=self.fmttext.split(" ")
        
        wordDict={}
        for w in set(listText):#chaange to set to remove duplicates
            wordDict[w]=listText.count(w)#loop through counting how many times a unique word appears in the original list
            
        return (wordDict)
    def freqof(self,word):
        freqDict=self.freqAll()
        if word in freqDict:
            return (freqDict[word])

        else:
            return(0)
```
